```python
import os
import re
import PyPDF2
import pandas as pd

def extract_text_and_dates(pdf_path):
    text_by_page = []
    release_dates = []
    course_pattern_years = []
    ledger_years = []
    with open(pdf_path, "rb") as pdf_file:
        pdf_reader = PyPDF2.PdfReader(pdf_file)
        current_course_pattern_year = None
        current_ledger_year = None
        for page in pdf_reader.pages:
            page_text = page.extract_text()

            # Updated release_date_pattern to capture both formats
            release_date_matches = re.findall(r'(?:(\d{1,2}/\w+/\d{4})|(\d{1,2}-\w+-\d{4}))', page_text)
            release_date = next((date for date in release_date_matches[0] if date), None)

            # Append the release_date to the list
            release_dates.append(release_date)

            course_pattern_match = re.search(r'B\.Arch \((\d{4} Pattern)\)', page_text)
            if course_pattern_match:
                current_course_pattern_year = course_pattern_match.group(1)
            course_pattern_years.append(current_course_pattern_year)

            ledger_year_match = re.search(r'(Supplementary|College) Ledger\(([^)]+)\)', page_text)
            if ledger_year_match:
                current_ledger_year = ledger_year_match.group(2)
            ledger_years.append(current_ledger_year)

            page_text = re.sub(r'Page \d+', '', page_text)
            text_by_page.append(page_text)

    return text_by_page, release_dates, course_pattern_years, ledger_years

def process_partitions(all_text, release_dates, course_pattern_years, ledger_years, dotted_line_pattern):
    all_partitions = re.split(dotted_line_pattern, all_text)
    current_page = 0
    student_data = []
    for partition in all_partitions:
        page_breaks = partition.count('\f')
        current_page += page_breaks
        partition = partition.replace('\f', '')

        student_pattern = r"PRN: (\d+).*?Seat no.: (\d+).*?NAME: (.*?)(?: Mother - (.*?))?\n"
        matches = re.finditer(student_pattern, partition, re.DOTALL)
        for match in matches:
            prn, seat, name, mother = match.groups()
            release_date = release_dates[current_page - page_breaks]
            course_pattern_year = course_pattern_years[current_page - page_breaks]
            ledger_year = ledger_years[current_page - page_breaks]

            student_info = {
                "PRN": str(prn),
                "Seat No": str(seat),
                "Name": str(name),
                "Mother Name": str(mother) if mother else None,
                "Course Pattern Year": course_pattern_year,
                "Ledger Year": ledger_year,
                "Ledger Release Date": release_date,
                "Student Info Lines": partition.strip().split('\n')
            }
            student_data.append(student_info)
    return student_data

pdf_directory = "/content/gdrive/MyDrive/BNCA Project August 2023/BNCA_Sample_Data_Files/New_Data_Files/All Files of 2019 Pattern "
pdf_files = [file for file in os.listdir(pdf_directory) if file.lower().endswith(".pdf")]

print("Total files for extraction are:", len(pdf_files))

all_student_records = []
dotted_line_pattern = r"\.{100,}"

for pdf_file_name in pdf_files:
    pdf_file_path = os.path.join(pdf_directory, pdf_file_name)
    pages_text, pages_release_dates, pages_course_pattern_years, pages_ledger_years = extract_text_and_dates(pdf_file_path)

    all_text = '\f'.join(pages_text)
    student_records = process_partitions(all_text, pages_release_dates, pages_course_pattern_years, pages_ledger_years, dotted_line_pattern)
    all_student_records.extend(student_records)
    print("Just cross-checking the files:", pdf_file_name)

All_P19_YearDateWise_Student_Info_df1923 = pd.DataFrame(all_student_records)
#print(All_P19_YearDateWise_Student_Info_df1923)
```

##**Final SGPA and Year Result For Four Ledgers**
One more approch to get Smester SGPA and Year Result together without any mistakes

* By refering root dataframe **All_P19_YearDateWise_Student_Info_df1923**
* Getting output as SGPA and Year Result together into **P19_YearDateWise_SGPA_YearResult_Info_df1923** With all validations like PRN, Course Year, Ledger Year and Semeters.

```python
import pandas as pd
import re

def parse_info(student_info_lines):
    info_str = ' '.join(student_info_lines)

    semester_data = []
    year_data = []

    semester_info_pattern = r'(\w+ Semester) SGPA\s*:\s*([-\d.]+) Credits Earned/Total\s*:\s*(\d+/\d+)'
    total_credit_points_pattern = r'Total Credit Points:\s*(\d+)'
    year_info_pattern_with_result = r'(First|Second|Third|Fourth|Fifth) Year Result\s*:\s*(.*?)\s*Total Credits Earned\s*:\s*(\d+/\d+)'
    year_info_pattern_without_result = r'(First|Second|Third|Fourth|Fifth) Year Total Credits Earned\s*:\s*(\d+/\d+)'

    semester_matches = re.finditer(semester_info_pattern, info_str)
    total_credit_points_matches = re.findall(total_credit_points_pattern, info_str)
    year_matches_with_result = re.finditer(year_info_pattern_with_result, info_str)
    year_matches_without_result = re.finditer(year_info_pattern_without_result, info_str)

    for match in semester_matches:
        sgpa = match.group(2).strip()
        sgpa = sgpa if sgpa.replace('.', '', 1).isdigit() else None
        semester_data.append({
            'SGPA Semester Name': match.group(1),
            'Semester SGPA': sgpa,
            'Semester Credits Earned': match.group(3)
        })

    for tcp, sem_data in zip(total_credit_points_matches, semester_data):
        sem_data['Semester Total Credit Points'] = tcp

    for match in year_matches_with_result:
        year_data.append({
            'Year Result Year Name': match.group(1) + " Year",
            'Year Result': match.group(2).strip(),
            'Year Total Credits Earned': match.group(3)
        })

    for match in year_matches_without_result:
        year_data.append({
            'Year Result Year Name': match.group(1) + " Year",
            'Year Result': None,
            'Year Total Credits Earned': match.group(2)
        })

    return semester_data, year_data

def match_year_result_to_semesters(semesters, years, prn, course_year, ledger_year):
    semester_to_year_map = {
        "First Semester": "First Year",
        "Second Semester": "First Year",
        "Third Semester": "Second Year",
        "Fourth Semester": "Second Year",
        "Fifth Semester": "Third Year",
        "Sixth Semester": "Third Year",
        "Seventh Semester": "Fourth Year",
        "Eighth Semester": "Fourth Year",
        "Ninth Semester": "Fifth Year",
        "Tenth Semester": "Fifth Year",
    }

    year_result_dict = {(year_info['Year Result Year Name'], prn, course_year, ledger_year): year_info
                        for year_info in years}

    for sem in semesters:
        year_key = (semester_to_year_map.get(sem['SGPA Semester Name'], ''), prn, course_year, ledger_year)
        if year_key in year_result_dict:
            sem.update(year_result_dict[year_key])

new_data = []

for index, row in All_P19_YearDateWise_Student_Info_df1923.iterrows():
    semester_info, year_info = parse_info(row['Student Info Lines'])

    match_year_result_to_semesters(semester_info, year_info, row['PRN'], row['Course Pattern Year'], row['Ledger Year'])

    for sem_data in semester_info:
        new_record = {
            'PRN': row['PRN'],
            'Seat No': row.get('Seat No', 'N/A'),
            'Name': row['Name'],
            'Course Pattern Year': row['Course Pattern Year'],
            'Ledger Year': row['Ledger Year'],
            'Ledger Release Date': row['Ledger Release Date'],
            **sem_data
        }
        new_data.append(new_record)

P19_YearDateWise_SGPA_YearResult_Info_df1923 = pd.DataFrame(new_data)
#print(P19_YearDateWise_SGPA_YearResult_Info_df1923.head())
```

```python
import pandas as pd
import re

def parse_info(cgpa_info_lines):
    info_str = ' '.join(cgpa_info_lines)

    semester_data = []
    year_data = []

    semester_info_pattern = r'(\w+ Semester) SGPA\s*:\s*([-\d.]+) Credits Earned/Total\s*:\s*(\d+/\d+)'
    total_credit_points_pattern = r'Total Credit Points:\s*(\d+)'
    year_info_pattern_with_result = r'(First|Second|Third|Fourth|Fifth) Year Result\s*:\s*(.*?)\s*Total Credits Earned\s*:\s*(\d+/\d+)'
    year_info_pattern_without_result = r'(First|Second|Third|Fourth|Fifth) Year Total Credits Earned\s*:\s*(\d+/\d+)'
    cgpa_class_pattern = r'C\.G\.P\.A\. : (\d+(\.\d+)?)\s*([\w\s]+?)(?=\d|$)'

    semester_matches = re.finditer(semester_info_pattern, info_str)
    total_credit_points_matches = re.findall(total_credit_points_pattern, info_str)
    year_matches_with_result = re.finditer(year_info_pattern_with_result, info_str)
    year_matches_without_result = re.finditer(year_info_pattern_without_result, info_str)
    cgpa_class_match = re.search(cgpa_class_pattern, info_str)

    for match in semester_matches:
        sgpa = match.group(2).strip()
        sgpa = sgpa if sgpa.replace('.', '', 1).isdigit() else None
        semester_data.append({
            'SGPA Semester Name': match.group(1),
            'Semester SGPA': sgpa,
            'Semester Credits Earned': match.group(3)
        })

    for tcp, sem_data in zip(total_credit_points_matches, semester_data):
        sem_data['Semester Total Credit Points'] = tcp

    for match in year_matches_with_result:
        year_data.append({
            'Year Result Year Name': match.group(1) + " Year",
            'Year Result': match.group(2).strip(),
            'Year Total Credits Earned': match.group(3)
        })

    for match in year_matches_without_result:
        year_data.append({
            'Year Result Year Name': match.group(1) + " Year",
            'Year Result': None,
            'Year Total Credits Earned': match.group(2)
        })

    cgpa = cgpa_class_match.group(1) if cgpa_class_match else None
    cgpa_info = {
        'C.G.P.A.': cgpa,
        'Class': cgpa_class_match.group(3).strip() if cgpa_class_match else None
    }

    return semester_data, year_data, cgpa_info

def match_year_result_to_semesters(semesters, years, cgpa_info, prn, course_year, ledger_year):
    semester_to_year_map = {
        "First Semester": "First Year",
        "Second Semester": "First Year",
        "Third Semester": "Second Year",
        "Fourth Semester": "Second Year",
        "Fifth Semester": "Third Year",
        "Sixth Semester": "Third Year",
        "Seventh Semester": "Fourth Year",
        "Eighth Semester": "Fourth Year",
        "Ninth Semester": "Fifth Year",
        "Tenth Semester": "Fifth Year",
    }

    year_result_dict = {(year_info['Year Result Year Name'], prn, course_year, ledger_year): year_info
                        for year_info in years}

    for sem in semesters:
        year_key = (semester_to_year_map.get(sem['SGPA Semester Name'], ''), prn, course_year, ledger_year)
        if year_key in year_result_dict:
            sem.update(year_result_dict[year_key])

new_data = []

# Assuming All_P19_YearDateWise_Student_Info_df1923 is the DataFrame containing the data
for index, row in All_P19_YearDateWise_Student_Info_df1923.iterrows():
    semester_info, year_info, cgpa_info = parse_info(row['Student Info Lines'])

    match_year_result_to_semesters(semester_info, year_info, cgpa_info, row['PRN'], row['Course Pattern Year'], row['Ledger Year'])

    for sem_data in semester_info:
        new_record = {
            'PRN': row['PRN'],
            'Seat No': row.get('Seat No', 'N/A'),
            'Name': row['Name'],
            'Course Pattern Year': row['Course Pattern Year'],
            'Ledger Year': row['Ledger Year'],
            'Ledger Release Date': row['Ledger Release Date'],
            **sem_data,
            **cgpa_info  # Include C.G.P.A. and Class info
        }
        new_data.append(new_record)

P19_YearDateWise_SGPA_YearResult_CGPA_Info_df1923 = pd.DataFrame(new_data)
#print(P19_YearDateWise_SGPA_YearResult_CGPA_Info_df1923.head())
```
